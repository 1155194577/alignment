```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
# Load the CSV file
df = pd.read_csv('second_search.csv')
# df.head()
df
```

```python
vertical_displacement = df['vertical_displacement'] 
cosine_similarity = df['cosine_similarity']
mse = df['mse']
mae = df['mae']
mean_difference = df['mean_difference']
```

```python
plt.plot(vertical_displacement, cosine_similarity, marker='o')
plt.xlabel('Vertical Displacement')
plt.ylabel('cosine_similarity')
plt.title('Vertical Displacement vs Cosine Similarity')
plt.grid(True)
plt.ylim(0, 1)
plt.show()
```

```python
plt.plot(vertical_displacement, mse, marker='o')
plt.xlabel('Vertical Displacement')
plt.ylabel('mse')
plt.title('mse vs Vertical Displacement')
plt.grid(True)
plt.show()

min_mse = mse.min()
min_mse_index = mse.idxmin()
corresponding_vertical_displacement = vertical_displacement[min_mse_index]
print("Minimum MSE:", min_mse)
print("Corresponding Vertical Displacement:", corresponding_vertical_displacement)
```

```python
plt.plot(vertical_displacement, mae, marker='o')
plt.xlabel('Vertical Displacement')
plt.ylabel('mae')
plt.title('mae vs Vertical Displacement')
plt.grid(True)
plt.show()

min_mae = mae.min()
min_mae_index = mae.idxmin()
corresponding_vertical_displacement_mae = vertical_displacement[min_mae_index]
print("Minimum MAE:", min_mae)
print("Corresponding Vertical Displacement:", corresponding_vertical_displacement_mae)
```

```python
plt.plot(vertical_displacement, mean_difference, marker='o')
plt.xlabel('Vertical Displacement')
plt.ylabel('mean_difference')
plt.title('difference of mean vs Vertical Displacement')
plt.grid(True)
plt.show()

min_mean_difference = mean_difference.min()
min_mean_difference_index = mean_difference.idxmin()
corresponding_vd_for_mean_diff = vertical_displacement[min_mean_difference_index]
print("Minimum mean_difference:", min_mean_difference)
print("Corresponding Vertical Displacement:", corresponding_vd_for_mean_diff)
```
